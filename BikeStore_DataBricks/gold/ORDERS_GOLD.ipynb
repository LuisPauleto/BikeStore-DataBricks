```sql
%python
df_orders_gold = spark.sql("""
select 
o.order_id,
o.customer_id,
o.city,
CAST(o.order_date AS DATE) AS order_date,
CAST(o.required_date as date) AS required_date,
o.order_status,
o.status,
o.store_id,
o.total_sale as sales_acumullated,
o.first_name as staff_name

from bikestore_project.silver.orders o

ORDER BY order_date desc
""")

display(df_orders_gold)
```

```sql
%python
df_orders_gold.write.mode("overwrite")\
    .format("delta")\
    .saveAsTable("bikestore_project.gold.orders")
```
